```python
from bs_risk import aad_greeks

S0 = 100
K = 103
V = 0.20
r = 0.01
T = 1.00

pv, *_, theta = aad_greeks(100, 103, 0.20, 0.01, 1.00)
dS = -3
dV = 0.02
dt = 1 / 252

# Roll to T COB and compute Greeks
new = S0 + dS, K, V + dV, r, T - dt
decay = S0, K, V, r, T - dt

pv_now, _, _, _, _ = aad_greeks(*new)
_, delta, vega, _, _ = aad_greeks(*decay)

gammaBump = 0.005 * S0
_, delta_up, _, _, _ = aad_greeks(S0 + gammaBump, K, V, r, T - dt)
_, delta_do, _, _, _ = aad_greeks(S0 - gammaBump, K, V, r, T - dt)
gamma = (delta_up - delta_do) / (2 * gammaBump)

attribution = delta * dS - theta * dt + 1 / 2 * gamma * (dS) ** 2 + vega * dV
pnl = pv_now - pv
100 * attribution, pnl * 100, (pnl - attribution) * 100
```
